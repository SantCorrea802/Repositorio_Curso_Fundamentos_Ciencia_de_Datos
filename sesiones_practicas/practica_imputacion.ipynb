```python
import pandas as pd
import numpy as np
import missingno as msno
from sklearn.impute import SimpleImputer
from difflib import get_close_matches
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.DataFrame({
    'Nombre': ['Ana', 'Luis', 'Pedro', None, 'Marta', 'Luis', 'Sofía'],
    'Edad': [25, np.nan, 35, 29, -5, 25, None],
    'Ciudad': ['Bogotá', 'Medellín', None, 'Medellín', 'Cali', 'Bogotá', 'Cali'],
    'Ingreso': [3500, 4800, np.nan, 5200, 5100, np.nan, 4700],
    'FechaIngreso': ['2023-01-01', '2023-01-05', None, '2023-01-10', '2023-01-12', None, '2023-01-15']
})
```

```python
df
```

```python
for col in df.columns:
  print(df[col].unique())
```

```python
df['Ingreso'] = df['Ingreso'].fillna(df['Ingreso'].median())
df['Nombre'] = df['Nombre'].fillna(df['Nombre'].mode()[0])
df['Ciudad'] = df['Ciudad'].fillna(df['Ciudad'].mode())
df['FechaIngreso'] = df['FechaIngreso'].fillna(df['FechaIngreso'].mode()[0])
df['FechaIngreso'] = pd.to_datetime(df['FechaIngreso'], errors='coerce')
df['Edad'] = df['Edad'].apply(lambda x: (df['Edad'].median()) if x < 0 else x)
df['Edad'] = df['Edad'].fillna(df['Edad'].median())
df
```

# TERCER PUNTO

```python
df = pd.DataFrame({
    'Ciudad': ['bogota', 'Bogotá', 'BOGOTA', 'bogotá', 'bogata', 'Bógota', 'BogoTa', 'Cali', 'calí', 'medellín', 'medellin']
})
df
```

```python
df["Ciudad"] = df["Ciudad"].str.lower()
df
```

```python
df["Ciudad"].iloc[7]
```

```python
df["Ciudad"].shape[0]
```

```python
admitidas = ["cali", "medellin", "bogota"]
```

```python
for i in range(df.shape[0]):
    ciudad = df.loc[i, "Ciudad"]


    if ciudad not in admitidas:
        match = get_close_matches(ciudad, admitidas, n=1, cutoff=0.7)
        if match:
            df.loc[i, "Ciudad"] = match[0]

df
```

```python
df.drop_duplicates(keep="last", inplace=True)
df
```
